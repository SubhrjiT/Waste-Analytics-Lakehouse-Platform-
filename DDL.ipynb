```python
%%sql
CREATE TABLE IF NOT EXISTS bronze.retail (
  Id           string,
  Date         date,
  Open         string,
  High         string,
  Low          string,
  Close        string,
  Volume       string,
  Stock_Trading string,
  createdBy    TIMESTAMP,
  Added_By     string,
  Added_On     timestamp
);
```
